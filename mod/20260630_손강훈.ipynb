```python
html_books = """
<div class='library'>
<div class='book' id='b-101'>
<h3 class='book-title'>딥러닝 텐서플로 마스터</h3>
<p class='book-author'>김작가</p>
</div>
<div class='book' id='b-102'>
<h3 class='book-title'>이것이 진짜 R 프로그래밍</h3>
<p class='book-author'>박통계</p>
</div>
<div class='book' id='b-103'>
<h3 class='book-title'>자연어 처리 입문</h3>
<p class='book-author'>최이과</p>
</div>
</div>
"""
```

```python
from bs4 import BeautifulSoup

# BeautifulSoup 객체 생성
soup = BeautifulSoup(html_books, "html.parser")

# book 클래스가 있는 div 모두 찾기
books = soup.find_all("div", class_="book")

books_info = []

# 책 정보 추출
for book in books:

    # 제목
    title = book.find("h3", class_="book-title").text

    # 저자
    author = book.find("p", class_="book-author").text

    # 딕셔너리 형태로 저장
    books_info.append({
        "title": title,
        "author": author
    })

print("=== 추출된 도서 목록 ===")
print(books_info)
```

```python
html_news = """
<ul class='news-section'>
<li class='news-item'><a href='https://news.example.com/ai-1'>인공지능 규제법 통과, IT 업계 파장</a></li>
<li class='news-item'><a href='https://news.example.com/dev-2'>파이썬 3.12 릴리즈, 무엇이 달라졌나?</a></li>
<li class='news-item'><a href='https://news.example.com/sec-3'>오픈소스 라이브러리 보안 취약점 경고</a></li>
</ul>
"""
```

```python
from bs4 import BeautifulSoup

# BeautifulSoup 객체 생성
soup = BeautifulSoup(html_news, "html.parser")

# 뉴스 목록 찾기
news = soup.find_all("li", class_="news-item")

news_list = []

# 뉴스 하나씩 반복
for item in news:

    # a 태그 찾기
    link = item.find("a")

    # 제목 추출
    headline = link.text

    # href 속성 추출
    url = link["href"]

    # 딕셔너리로 저장
    news_list.append({
        "headline": headline,
        "url": url
    })

print("=== 수집된 뉴스 헤드라인 ===")
print(news_list)
```

```python
import sqlite3

# 메모리 DB 연결
conn = sqlite3.connect(":memory:")

# Cursor 생성
cursor = conn.cursor()

# users 테이블 생성
cursor.execute("""
CREATE TABLE users(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    email TEXT,
    level TEXT
)
""")

# 회원 정보 입력
cursor.execute(
    "INSERT INTO users(name, email, level) VALUES(?, ?, ?)",
    ("홍길동", "hong@example.com", "gold")
)

cursor.execute(
    "INSERT INTO users(name, email, level) VALUES(?, ?, ?)",
    ("이영희", "lee@example.com", "silver")
)

cursor.execute(
    "INSERT INTO users(name, email, level) VALUES(?, ?, ?)",
    ("김철수", "kim@example.com", "gold")
)

# 저장
conn.commit()

# gold 회원 조회
cursor.execute(
    "SELECT name, email FROM users WHERE level = ?",
    ("gold",)
)

# 결과 저장
gold_users = cursor.fetchall()

print("=== Gold 등급 회원 목록 ===")
print(gold_users)

# 연결 종료
conn.close()
```

```python
import pandas as pd
import sqlite3

# 실습 데이터프레임 생성
df_members = pd.DataFrame({
    "member_id": ["M01", "M02", "M03"],
    "name": ["이민수", "김영희", "박준형"],
    "score": [85, 95, 78]
})

# DB 연결
conn = sqlite3.connect("users.db")

# DataFrame을 members 테이블로 저장
df_members.to_sql(
    "members",
    conn,
    if_exists="replace",
    index=False
)

# 저장
conn.commit()

# 연결 종료
conn.close()

# 다시 DB 연결
conn = sqlite3.connect("users.db")

# SQL로 데이터 읽기
df_read = pd.read_sql_query(
    "SELECT * FROM members",
    conn
)

# 연결 종료
conn.close()

print("=== DB 복원 데이터프레임 ===")
print(df_read)
```
