```python
# 사칙연산을 하는 클래스 => 클래스 이름, 변수, 함수(메서드), 초기화 함수(생성자)
# 클래스 이름은 대문자로 시작
# 캐멀케이스
# 변수 : 두개
# 함수(메서드) : 설정, +, -, *, /
class Fourcal :
    # 클래스의 멤버 : 변수, 메서드
    # 메서드를 통해 변수()를 생성 => 클래스를 통해서 객체로 만들면 변수가 메모리에 로드됌
    def setData(self, first, second): # self : 현재 인스턴스 멤버를 사용하기 위함
        self.first = first
        self.second = second

    def printData(self):
        print(f'first : {self.first}, second : {self.second}')

    def add(self):
        return self.first + self.second
    
    def sub(self):
        return self.first - self.second

    def mul(self):
        return self.first * self.second
    
    def div(self):
        return self.first / self.second
```

```python
cal1 = Fourcal()
```

```python
print(id(cal1), type(cal1))
```

```python
cal1.setData(10,7)
```

```python
print(cal1.first, cal1.second)
cal1.printData()
```

```python
print(cal1.first, cal1.second)
cal1.printData()
print(cal1.add())
print(cal1.sub())
print(cal1.mul())
print(cal1.div())
```
