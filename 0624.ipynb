```python
# 반복구문 

#1~5 까지 숫자를 출력

cnt = 1 # 초기값
while cnt < 6 :
    print(0)
    cnt = cnt + 1
```

```python
#continue
#반복문 내부에서 사용 (조건 적)
#홀수 출력

cnt = 0
while cnt < 10 :
    cnt += 1
    if cnt%2 == 0:
        continue
    print(cnt)
```

```python
# break : 반복문 내에서 사용 => 이 키워드를 만나면 반복문 종료
cnt = 0
while True :
    # 조건식을 두고 break
    cnt += 1
    print(cnt) 
    if cnt > 4 :
        break
    
```

```python
# 5명의 삭행이 영어 시험 점수를 리스트로 저장
# 학생별로 60점 이상이면 PASS 출력, 아니면 FAIL 출력
# 총점과 평균 점수 출력

scores = [80, 50, 70, 100, 55]
snum = 0
sum = 0

for s in scores : 
    snum += 1
    check = 'PASS' if s>60 else 'FAIL'
    print(snum, s, check)
    sum = sum + s

print('-'*30)
print('총점:', sum)
print('평균:', sum/len(scores))
    
```

```python
# 함수 : 데이터를 받고 처리해서 결과를 반환하는 코드들의 집합
# 처리를 위한 코드가 필요하면 호출해서 사용
# 가독성, 유지보수, 확장성

# def 함수이름(매개변수, 매개변수, 매개변수, .....) :
#   처리할 코드들 
#      return 데이터
```

```python
# 숫자 두개를 받아서 더하기 연산을 한 후 연산의 결과 값을 반환하는 함수

def add(n1, n2):
    result = n1 + n2
    return result
```

```python
# 함수 호출
add(10, 20)
```

```python
# 함수도 객체다
print(type(add))
```

```python
# 가변매개변수
# 받아야 하는 매개변수의 숫자를 모를때! => 전달되는 모든 매개변수를 tuple로 묶어서 사용

# 전달되는 모든 매개변수(숫자)를 모두 더한 값을 반환 하는 함수

def add_all_sum(*args):
    print(args)
    print(type(args))
```

```python
def proc_type(proc_type, *args):
    print(proc_type, args)
    if proc_type=='+':
        result = 0
        for num in args:
            result += num
        return result
    elif proc_type=='*':
        result = 1
        for num in args :
            result *= num
        return result
    else :
        print('연산이 가능한 연산기호를 입력하세요! ( + | *)')
```

```python
proc_type('*', 1,2,3,4,5)
```
