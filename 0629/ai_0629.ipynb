```python
# pandas를 이용 => 평균, 중앙, 최빈
import pandas as pd
```

```python
data = [10, 15, 10, 20, 30, 10, 40]
data_serises = pd.Series(data)
```

```python
mean_val = data_serises.mean() #평균값
median_val = data_serises.median()  #중간값
mode_val = data_serises.mode()[0]  #최반값
```

```python
print(mean_val)
print(median_val)
print(mode_val)
```

```python
data = {'점수': [55, 60, 65, 70, 75, 80, 85, 90, 95]}
```

```python
df = pd.DataFrame(data)
```

```python
# 분산
var_val = df['점수'].var()
print(var_val)
```

```python
# 표준편차
std_val = df['점수'].std()
print(std_val)
```

```python
# 사분위 점수
q1_val = df['점수'].quantile(0.25) # 1사분위
q2_val = df['점수'].quantile(0.5)  # 2사분위
q3_val = df['점수'].quantile(0.75) # 3사분위
q4_val = df['점수'].max()          # 4사분위
```

```python
print(q1_val, q2_val, q3_val, q4_val)
```

```python
print('IQR', q3_val - q1_val)
```

```python
df_1 = df[['번호', '영업상태명', '사업장명', '소재지전체주소', '도로명전체주소']]
df_1
```

```python
# 소재지 기준 값 => NaN의 행을 삭제
df_2 = df_1.dropna(subset=['소재지전체주소'])
df_2
```

```python
# 소재지 기준 => '서울특별시' 포함하는 행만 남기는 처리
df_3 = df_2[df_2['소재지전체주소'].str.contains('서울특별시')]
df_3
```

```python
df_3.info()
```

```python
df_4 = df_3[df_3['사업장명'].str.contains('스타벅스') ]
df_4
```

```python
df_5 = df_4[df_4['영업상태명'] !='폐업']
df_4
```

```python
'서울특별시 마포구 대흥동 111-1'.split()
```

```python
# 열 추가
df_5.loc[:, '시군구'] = df_5['소재지전체주소'].str.split().str[1]
df_5
```

```python
# 그룹핑
df_6 = df_5.groupby('시군구').size()
df_6
```

```python
#  차트
#plt.rcParams['font.family'] = 'Batang'
plt.bar(
    df_6['시군구'],
    df_6['사업장수']
)

plt.xticks(rotation=45)
plt.title('서울시 스아벅스 사업장 분석표')
plt.xlabel('시군구')
plt.ylabel('사업장수')
```
